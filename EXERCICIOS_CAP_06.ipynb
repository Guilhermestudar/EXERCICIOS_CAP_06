## Exercício 6.1: Hastes para Transmissões
**Contexto:** O fabricante deseja monitorar o processo de produção de hastes para transmissões automotivas. Foram coletadas 25 amostras de tamanho $n=4$ para avaliar a estabilidade do diâmetro interno.

**Dados do Problema:**
* Número de amostras ($k$): 20
* Tamanho da amostra ($n$): 4
* Somatório das médias ($\sum \bar{x}_i$): 10.275
* Somatório das amplitudes ($\sum R_i$): 1,012

**Objetivo:** Calcular os limites de controle para os gráficos $\bar{X}$ e $R$ e estimar os parâmetros do processo ($\mu$ e $\sigma$), assumindo que o processo está sob controle.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Tabela de Constantes (Apêndice VI)
constantes = {
    4: {'A2': 0.729, 'D3': 0, 'D4': 2.282, 'd2': 2.059},
    5: {'A2': 0.577, 'D3': 0, 'D4': 2.114, 'd2': 2.326}
}
```

```python
# Cálculo dos Parâmetros
k, n = 20, 4
X_db = 10275 / k
R_b = 1.012 / k

# Constantes para n=4 (Apêndice VI)
A2 = constantes[4]['A2']
D3 = constantes[4]['D3']
D4 = constantes[4]['D4']
d2 = constantes[4]['d2']

# Limites X-barra
LSC_X = X_db + (A2 * R_b)
LIC_X = X_db - (A2 * R_b)

# Limites R
LSC_R = D4 * R_b
LIC_R = D3 * R_b

print(f"X-Barra: LSC={LSC_X:.4f}, LC={X_db:.4f}, LIC={LIC_X:.4f}")
print(f"R: LSC={LSC_R:.4f}, LC={R_b:.4f}, LIC={LIC_R:.4f}")
```

## **Análise de Resultados:**
Os limites de controle calculados definem a faixa de variação esperada para causas comuns do processo. Qualquer amostra cujos valores de $\bar{X}$ ou $R$ estejam fora desses limites sugere a presença de uma causa atribuível de variação, exigindo investigação corretiva. A estimativa do desvio-padrão do processo ($\hat{\sigma} = \bar{R}/d_2$) é de {sigma_est:.4f}, servindo como base para futuros cálculos de capacidade do processo.

## Análise de Padrões (Regras de Shewhart)
Esta célula deve ser executada antes de iniciar os exercícios, pois define as regras básicas de detecção de causas especiais.

```python
# Função para verificar regras de controle básicas (Western Electric/Montgomery)
def verificar_regras(series):
    # Regra 1: Ponto fora dos limites (exemplo simplificado)
    # Regra 2: 9 pontos consecutivos no mesmo lado da média
    violacoes = []
    media = series.mean()
    lados = series > media
    for i in range(len(series) - 8):
        if all(lados[i:i+9]) or all(~lados[i:i+9]):
            violacoes.append(f"Regra de 9 pontos violada na amostra {i+9}")
    return violacoes
```

### Exercício 6.18: Desempenho e Sucata
**Problema:** $k=50$ amostras de $n=6$. [cite_start]A soma de $\bar{X}$ é 2.000 e a soma de $R$ é 200[cite: 3632, 3634]. [cite_start]Além de calcular os limites, precisamos descobrir a porcentagem de sucata (abaixo de 36) e retrabalho (acima de 46) com base nas especificações $41 \pm 5,0$ [cite: 3639, 3641-3644].

```python
k, n = 50, 6
X_db = 2000 / k
R_b = 200 / k
c6 = constantes[n]

print(f"(a) Limites X-barra: LSC={X_db + c6['A2']*R_b:.3f} | LIC={X_db - c6['A2']*R_b:.3f}")

sigma_est = R_b / c6['d2']
print(f"(b) Limites naturais (3 sigmas): {X_db - 3*sigma_est:.3f} a {X_db + 3*sigma_est:.3f}")

# (d) Fração de sucata (abaixo de 36) e retrabalho (acima de 46)
LSE, LIE = 46.0, 36.0
p_sucata = norm.cdf(LIE, X_db, sigma_est)
p_retrabalho = 1 - norm.cdf(LSE, X_db, sigma_est)

print(f"(d) Sucata gerada: {p_sucata*100:.2f}%")
print(f"(d) Retrabalho gerado: {p_retrabalho*100:.2f}%")
print(f"(e) Sugestão: A média atual é {X_db}, mas o alvo da especificação é 41. O processo deve ser recentralizado para 41.")
```

1. A média atual do processo é $\bar{\bar{x}} = 40$, mas o alvo das especificações é $41$.
2. A distribuição da curva normal nos mostra que, devido a esse deslocamento da média para a esquerda, o processo está gerando mais **sucata** (peças abaixo de 36) do que **retrabalho** (peças acima de 46).
3. **Recomendação Técnica:** A máquina não precisa ser substituída, mas deve ser **recentralizada**. Ajustar a configuração da máquina para que a média passe de 40 para o alvo de 41 equilibrará as caudas da distribuição, minimizando drasticamente a soma total de itens defeituosos (sucata + retrabalho), sem precisar alterar a variância ($\sigma$) do processo.

**Contexto:** O processo operava com $n=5$. Com base nesses limites históricos, a empresa deseja saber como os limites seriam alterados caso ela decidisse reduzir as amostras para $n=3$ (Ex. 6.23) para economizar na inspeção, ou aumentar para $n=8$ (Ex. 6.24) para ganhar poder estatístico.
**Teoria:** A estimativa do desvio-padrão da população ($\hat{\sigma} = \bar{R}/d_2$) baseada nos dados antigos é a constante universal do processo. Usamos esse $\hat{\sigma}$ com as novas constantes de $n_{novo}$ para recalcular os novos limites de $\bar{X}$ e $R$.

```python
# (Substitua X_db e R_b pelos valores calculados do Ex. 6.7 original.
# Aqui usaremos valores genéricos para ilustrar a transição)
X_db_antigo = 0.3015 # Exemplo (Ex. 6.7 usa nominal 0.3015)
R_b_antigo = 0.00154
n_antigo = 3 # O Ex. 6.7 na verdade diz "baseiam-se em subgrupos de tamanho 3"
# Nota: O enunciado do 6.23 no PDF fala "usando n=5", mas o 6.7 usou n=3.
# Vamos assumir que os parâmetros fornecidos são para n_antigo = 5 para prosseguir com a demonstração.

sigma_est = R_b_antigo / constantes[n_antigo]['d2']

def recalcular_limites(n_novo, sigma, mu):
    c = constantes[n_novo]
    R_b_novo = c['d2'] * sigma
    LSC_x = mu + c['A2'] * R_b_novo
    LIC_x = mu - c['A2'] * R_b_novo
    LSC_R = c['D4'] * R_b_novo
    LIC_R = c['D3'] * R_b_novo
    return LSC_x, LIC_x, LSC_R, LIC_R, R_b_novo

# Ex 6.23: Reduzindo para n=3
LSCx3, LICx3, LSCR3, LICR3, Rb3 = recalcular_limites(3, sigma_est, X_db_antigo)
print("--- Exercício 6.23 (n=3) ---")
print(f"Novos limites de X: LSC={LSCx3:.5f} | LIC={LICx3:.5f}")
print(f"Novos limites de R: LSC={LSCR3:.5f} | LC={Rb3:.5f}\n")

# Ex 6.24: Aumentando para n=8
LSCx8, LICx8, LSCR8, LICR8, Rb8 = recalcular_limites(8, sigma_est, X_db_antigo)
print("--- Exercício 6.24 (n=8) ---")
print(f"Novos limites de X: LSC={LSCx8:.5f} | LIC={LICx8:.5f}")
print(f"Novos limites de R: LSC={LSCR8:.5f} | LC={Rb8:.5f}")
```

**Análise dos Exercícios 6.23 e 6.24:** Esses exercícios demonstram o impacto do tamanho do subgrupo racional na sensibilidade do gráfico.
* **Para $n=3$ (Ex 6.23):** A largura dos limites de controle do gráfico $\bar{X}$ aumenta e o valor esperado para a amplitude cai. Embora a inspeção fique mais barata, a curva Característica de Operação (CO) revelará que o poder para detectar deslocamentos de $\approx 2\sigma$ na média cai significativamente, retardando a percepção de instabilidade.
* **Para $n=8$ (Ex 6.24):** Os limites de $\bar{X}$ se estreitam (ficam mais rigorosos), pois a média de 8 unidades tem menor variabilidade que a média de 5. Isso aumenta enormemente a probabilidade do gráfico detectar imediatamente a falha no processo, tornando-o estatisticamente superior, embora o custo de coleta de dados encareça na prática diária da fábrica.

## Exercícios 6.26 e 6.27: Gráficos de Atributos (p e np)
**Contexto:** Ao contrário dos gráficos anteriores, aqui monitoramos a fração de itens defeituosos ($p$) ou o número total de defeituosos ($np$). São ideais para inspeções visuais ("passa/não passa").
**Problema:** Monitoramento de um processo de montagem de placas eletrônicas onde $n=100$ e coletamos 20 amostras.

```python
# Dados das 20 amostras de tamanho n=100 (Tabela 6E.14)
# Número de defeituosos em cada amostra
defeitos = [12, 8, 6, 9, 10, 12, 14, 10, 8, 7, 10, 11, 9, 12, 8, 7, 9, 10, 11, 12]
n = 100
k = len(defeitos)

# Cálculo da fração média (p-barra)
p_barra = sum(defeitos) / (k * n)

# Limites do gráfico p (p-barra +/- 3 * raiz(p*(1-p)/n))
LSC_p = p_barra + 3 * np.sqrt((p_barra * (1 - p_barra)) / n)
LIC_p = max(0, p_barra - 3 * np.sqrt((p_barra * (1 - p_barra)) / n))

print(f"6.26 - Gráfico p: LSC={LSC_p:.4f}, LC={p_barra:.4f}, LIC={LIC_p:.4f}")
```

**Análise:** O gráfico $p$ é fundamental para monitorar a "frações de defeituosos". Se a fração média ($p$) estiver aumentando ao longo do tempo, isso sinaliza uma degradação do processo que o gráfico de variáveis (como diâmetro) pode não ter detectado inicialmente.

## Exercícios 6.28 e 6.29: Contagem de Defeitos (Gráficos c e u)
**Contexto:** Enquanto o gráfico $p$ conta *unidades* defeituosas, os gráficos $c$ e $u$ contam o *número total de defeitos* em uma amostra.
* **Gráfico c:** Usado quando a área de oportunidade é constante.
* **Gráfico u:** Usado quando a área de oportunidade varia (ex: áreas diferentes de tecido ou comprimentos diferentes de cabo).

```python
# Exemplo para Gráfico c (Número total de defeitos em 20 amostras)
c_valores = [2, 1, 3, 0, 4, 2, 2, 1, 0, 3, 2, 1, 1, 2, 0, 4, 1, 2, 1, 0]
c_barra = np.mean(c_valores)

LSC_c = c_barra + 3 * np.sqrt(c_barra)
LIC_c = max(0, c_barra - 3 * np.sqrt(c_barra))

print(f"6.28 - Gráfico c: LSC={LSC_c:.4f}, LC={c_barra:.4f}, LIC={LIC_c:.4f}")
```

**Análise:** O uso do gráfico $c$ é ideal para a contagem de defeitos em produtos complexos. Se o gráfico $c$ mostrar um ponto acima do LSC, significa que um "cluster" de defeitos ocorreu naquele lote, o que frequentemente aponta para uma falha específica em um lote de matéria-prima ou um erro operacional humano localizado.
